```python
from collections import Counter
import re
import matplotlib.pyplot as plt

# Load cleaned transcript
with open("Speaker_0025_00001_cleaned.txt", "r") as f:
    text = f.read().lower()

# Clean text
text = re.sub(r"[^a-z\s]", "", text)

stopwords = {
    "the","and","is","in","to","of","a","for","on","that","this",
    "with","as","are","was","were","be","by","it","from","or","at"
}

words = [w for w in text.split() if w not in stopwords and len(w) > 3]

word_freq = Counter(words).most_common(10)

# Separate words and counts
labels = [w for w, c in word_freq]
counts = [c for w, c in word_freq]

# Plot
plt.figure()
plt.bar(labels, counts)
plt.title("Top 10 Word Frequencies")
plt.xlabel("Words")
plt.ylabel("Frequency")
plt.show()
```
