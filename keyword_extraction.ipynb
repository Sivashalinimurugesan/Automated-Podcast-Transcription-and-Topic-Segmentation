```python
from collections import Counter
import re

# Load cleaned transcript
with open("Speaker_0025_00001_cleaned.txt", "r") as f:
    text = f.read().lower()

# Remove punctuation
text = re.sub(r"[^a-z\s]", "", text)

# Basic stopwords
stopwords = {
    "the","and","is","in","to","of","a","for","on","that","this",
    "with","as","are","was","were","be","by","it","from","or","at"
}

# Tokenize and filter
words = [w for w in text.split() if w not in stopwords and len(w) > 3]

# Count word frequency
word_freq = Counter(words)

# Top 10 keywords
top_keywords = word_freq.most_common(10)

print("Top Keywords:")
for word, freq in top_keywords:
    print(f"{word} : {freq}")

```
