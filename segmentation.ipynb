```python
# Load cleaned transcript
with open("Speaker_0025_00001_cleaned.txt", "r") as f:
    text = f.read()

# Simple segmentation based on sentence count
sentences = text.split(".")
segments = []

segment_size = 5  # group every 5 sentences

for i in range(0, len(sentences), segment_size):
    segment = ".".join(sentences[i:i+segment_size])
    segments.append(segment.strip())

# Display first few segments
for idx, seg in enumerate(segments[:3]):
    print(f"\n--- Segment {idx+1} ---\n")
    print(seg)
```
